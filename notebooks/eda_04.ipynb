```python
import pandas as pd
import numpy as np
import missingno as msno
from pathlib import Path

curr_dir = Path.cwd()
proj_root = curr_dir.parent
raw_data = proj_root / "data/raw/uci-secom.csv"

df = pd.read_csv(raw_data)
cols_to_drop = set()
```

```python
# To drop zero variance column & full null column
nan_cols = set(df.columns[df.isna().all()].tolist())
const_cols = set([c for c in df.columns if df[c].nunique() <= 1])

cols_to_drop.update(nan_cols)
cols_to_drop.update(const_cols)

print("Current Columns to drop: {}".format(len(cols_to_drop)))
```
